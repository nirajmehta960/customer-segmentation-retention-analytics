# Notebook 03: RFM Analysis
Compute RFM scores, assign rule-based segments, generate segment profiles, and save all EDA charts.

**Input:** `data/preprocessed/online_retail_preprocessed.csv`  
**Output:** `data/featured/customer_rfm.csv`, `data/featured/segment_profiles.csv`, `images/eda_charts/`

```python
import os
import sys
import pandas as pd
if os.path.basename(os.getcwd()) == 'notebooks':
    os.chdir('..')
sys.path.insert(0, '.')

from src.rfm_analysis import run_rfm_pipeline

df = pd.read_csv('data/preprocessed/online_retail_preprocessed.csv')
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
rfm = run_rfm_pipeline(df=df, save_path='data/featured/customer_rfm.csv')
rfm.head(10)
```

```python
# Segment profile summary
profile = rfm.groupby('segment').agg(
    customers=('CustomerID', 'count'),
    avg_recency=('recency', 'mean'),
    avg_frequency=('frequency', 'mean'),
    avg_monetary=('monetary', 'mean'),
    total_revenue=('monetary', 'sum')
).round(1)
profile['pct_revenue'] = (profile['total_revenue'] / profile['total_revenue'].sum() * 100).round(1)
profile
```

```python
# Generate and save all EDA charts to images/eda_charts/
from src.visualizations import generate_all_eda_charts

generate_all_eda_charts(df, rfm, save_dir='images/eda_charts')
```
